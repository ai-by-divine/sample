```python
import pandas as pd
```

```python

# Sample data
df = pd.DataFrame({
    'City': ['Lagos', 'Abuja', 'Lagos', 'PortHarcourt', 'Abuja'],
    'Bought': [1, 0, 1, 0, 1]
})

# Calculate mean per category
mean_per_city = df.groupby('City')['Bought'].mean()
mean_per_city

```

```python

# Map means back to the City column
df['City_encoded'] = df['City'].map(mean_per_city)
df
```
